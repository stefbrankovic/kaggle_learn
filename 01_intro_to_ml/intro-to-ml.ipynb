## 3. Your First Machine Learning Model
---
As we can see, there are 81 columns - too many for simple analysis. The question is how to choose relevant columns i.e. selecting right data for modeling.

To subset the data, we need to select the **prediction target** and **features**.

First, let's look at all dataset columns.

```python
home_data.columns
```

Clearly, the **prediction target** is SalePrice.

And for the **features**, we will choose LotArea, YearBuilt, 1stFlrSF, 2ndFlrSF, FullBath, BedroomAbvGr and TotRmsAbvGrd.

```python
y = home_data.SalePrice
features = ['LotArea', 'YearBuilt', '1stFlrSF', '2ndFlrSF', 'FullBath', 'BedroomAbvGr', 'TotRmsAbvGrd']

X = home_data[features]
```

Let's get a closer look on the data chosen for modeling. Again, we will use descrbibe() and head() methods from pandas.

```python
X.describe()
```

```python
X.head()
```

**Model building**.

There are four steps to building and using a model:
- **Define model type and parameters**
- **Model fitting**
- **Making predictions**
- **Model evaluation**: determining how accurate the predictions are.

For this problem, we will use <u>DecisionTreeRegressor</u> from sklearn.tree

```python
# Model specyfing and fitting
from sklearn.tree import DecisionTreeRegressor

iowa_model = DecisionTreeRegressor(random_state = 1) # specifying a model, random_state - seed
iowa_model.fit(X, y) # fitting the model
```

```python
# Making predictions
predictions = iowa_model.predict(X) # X - our data for model fitting
print(predictions)
```

Let's compare the predictions to the actual values for the first five rows.

```python
print("First five houses:")
head_houses = X.head()
print("Predicted prices:")
print(iowa_model.predict(head_houses))
print("Real prices:")
print(y.head())
```

## 4. Model Validation
---

Before, we used all the train data to fit our model, and we also evaluated the model on the test data. This is not a good practice, because the model has already seen all the train data, and it will work well for the data it was trained on, but it will perform badly for the new data - this is called **overfitting**.

Therefore, we will split our data on two subsets - one for training, and the other one for validation.

```python
# train_test_split function splits the data
from sklearn.model_selection import train_test_split

train_X, val_X, train_y, val_y = train_test_split(X, y, random_state = 1)
```

Now, we will repeat the model specifying and fitting  process, but **only on the train data** subset. 

Similarly, we will make predictions **only on the validation data** subset.

```python
iowa_model = DecisionTreeRegressor(random_state=1)
iowa_model.fit(train_X,train_y)
```

```python
val_predictions = iowa_model.predict(val_X)
```

Now, let's compare new predictions to the actual values.

```python
print("Predicted values:")
print(val_predictions[:5])
print("Real values:")
print(val_y.head())
```

To evaluate the model, we need to use some metrics. **Mean Absolute Error** gives us the mean value of absolute errors for all predictions, i.e. "how for" are our predictions on average.

```python
from sklearn.metrics import mean_absolute_error

val_mae = mean_absolute_error(val_y, val_predictions)
print(val_mae)
```
